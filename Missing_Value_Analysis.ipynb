```python
# Task 21: Missing Value Analysis

## Objective
Identify missing values in a dataset, calculate column-wise
missing-value counts and percentages, and identify columns
that require missing-data treatment.

## Deliverables
- Missing-value summary
- Column-wise missing-value counts
- Missing-value percentages
- Identification of columns requiring treatment
- Explanation of missing-data observations
```

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "student_id": [
        1001, 1002, 1003, 1004, 1005, 1006,
        1007, 1008, 1009, 1010, 1011, 1012
    ],

    "name": [
        "Aarav", "Diya", "Kabir", "Ananya",
        "Ishaan", "Meera", "Rohan", "Sara",
        "Vivaan", "Aditi", "Arjun", "Nisha"
    ],

    "department": [
        "Computer Science", "Electronics", "Mechanical",
        "Computer Science", "Civil", "Electronics",
        "Mechanical", "Computer Science", "Civil",
        "Electronics", "Mechanical", np.nan
    ],

    "age": [
        20, 21, np.nan, 22, 21, 20,
        22, 21, 20, 22, 21, 20
    ],

    "marks": [
        86, np.nan, 74, 95, 68, 88,
        np.nan, 93, 72, 84, 77, 90
    ],

    "attendance": [
        92, 88, 79, np.nan, 72, 90,
        85, 94, np.nan, 86, 75, 91
    ]
}

df = pd.DataFrame(data)

print("Student dataset created successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

display(df)
```

```python
# Save the dataset as a CSV file
df.to_csv("missing_values_dataset.csv", index=False)

# Load the CSV file using Pandas
df = pd.read_csv("missing_values_dataset.csv")

print("CSV dataset loaded successfully!")

display(df.head(12))
```

```python
print("MISSING VALUE BOOLEAN CHECK")

missing_boolean = df.isnull()

display(missing_boolean)
```

```python
print("MISSING VALUE CHECK USING isna()")

display(df.isna())
```

```python
print("MISSING VALUES IN EACH COLUMN")

missing_counts = df.isnull().sum()

display(missing_counts.to_frame(
    name="Missing Value Count"
))
```

```python
print("MISSING VALUE PERCENTAGES")

missing_percentage = (
    df.isnull().sum() / len(df)
) * 100

display(
    missing_percentage.round(2).to_frame(
        name="Missing Percentage"
    )
)
```

```python
missing_summary = pd.DataFrame({
    "Column": df.columns,

    "Data Type": [
        str(df[column].dtype)
        for column in df.columns
    ],

    "Total Records": len(df),

    "Missing Count": [
        df[column].isnull().sum()
        for column in df.columns
    ],

    "Missing Percentage": [
        round(df[column].isnull().mean() * 100, 2)
        for column in df.columns
    ],

    "Treatment Required": [
        df[column].isnull().any()
        for column in df.columns
    ]
})

print("COMPLETE MISSING VALUE SUMMARY")

display(missing_summary)
```

```python
columns_to_treat = missing_summary[
    missing_summary["Treatment Required"] == True
]

print("COLUMNS REQUIRING MISSING VALUE TREATMENT")

display(columns_to_treat)

print("\nColumn names requiring treatment:")

print(columns_to_treat["Column"].tolist())
```

```python
print("ROWS CONTAINING MISSING VALUES")

missing_records = df[df.isnull().any(axis=1)]

display(missing_records)

print(
    "\nTotal rows containing missing values:",
    len(missing_records)
)
```

```python
total_cells = df.size

total_missing = df.isnull().sum().sum()

total_present = total_cells - total_missing

overall_missing_percentage = (
    total_missing / total_cells
) * 100

print("=" * 55)
print("        OVERALL MISSING VALUE ANALYSIS")
print("=" * 55)

print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Total cells:", total_cells)
print("Present values:", total_present)
print("Missing values:", total_missing)

print(
    "Overall missing percentage:",
    round(overall_missing_percentage, 2), "%"
)
```

```python
treatment_recommendations = {
    "department": "Mode or Unknown category",
    "age": "Median imputation",
    "marks": "Median imputation",
    "attendance": "Median imputation"
}

print("MISSING VALUE TREATMENT RECOMMENDATIONS")

for column, recommendation in treatment_recommendations.items():
    if df[column].isnull().any():
        print(f"{column}: {recommendation}")
```

```python
missing_summary.to_csv(
    "missing_value_summary.csv",
    index=False
)

print("Missing-value summary exported successfully!")
```
